```python
import math

pred = [3, 5, 2, 7, 8, 4, 6, 9, 5, 10]

y_test = [2, 5, 3, 8, 7, 4, 5, 10, 6, 9]

n = len(y_test)
print(n)
```

```python
# 오차 10개를 저장
errors = []

for i in range(n):
    d = pred[i] - y_test[i]
    errors.append(d)

print(errors)
```

```python
absolute_errors = []
for e in errors:
    absolute_errors.append(abs(e))

print(absolute_errors)
```

```python
squared_errors = []
for error in errors:
    squared_errors.append(error ** 2)

mse = sum(squared_errors) / n
```

```python
rmse = math.sqrt(mse)
print('RMSE : ', rmse)
```

```python
# R²: 1 - (모델의 제곱 오차 합 / 정답 평균만 사용했을 때의 제곱 오차 합)
y_mean = sum(y_test) / n

ss_res = 0  # 모델 예측의 제곱 오차 합
ss_tot = 0  # 정답 평균을 기준으로 한 제곱 오차 합
```

```python
for i in range(n):
    ss_res += (y_test[i] - pred[i]) ** 2
    ss_tot += (y_test[i] - y_mean) ** 2

r2 = 1 - (ss_res / ss_tot)

print(r2)
```
