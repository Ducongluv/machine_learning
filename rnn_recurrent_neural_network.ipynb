```python
# numpy reshape() 메서드 : 차원축소, 차원확대 메서드

import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, SimpleRNN, Dense
```

```python
# 시계열 데이터를 예측 -> 학습에 사용하는 변수는 하나 -> 슬라이딩 윈도우 알고리즘을 사용해서 X n개, y 1개 분할 -> 인덱스 순서를 변경없이 사용해야 함.

# 1. 추가 데이터 (가상 데이터)
price = np.array([
    100, 102, 101, 105, 107, 106, 110, 112, 115, 113,
    118, 120, 119, 123, 125, 128, 126, 130, 133, 135
], dtype=float)

print(price)
print(price.shape)
```

```python
# 스케일링 -> 신경망 -> 무조건 한다 -> 왜? 큰수보다 작은수의 좀 더 학습이 잘 이루어짐
print(price.max())
price = price/price.max()
print(price)

# min-max 방식으로 정규화하면 0~1 값을 취할 수 있음

price2 = (price - price.min()) / (price.max() - price.min())
print(price2)
```

```python
# price 데이터셋 (20, ) -> X : (17, 3), y : (17, )

X, y = [], []
window_size = 3

for i in range(0, len(price2) - window_size):
    X.append(price[ i : i + window_size])
    y.append(price[ i + window_size])

print(X)
print(y)

```

```python
# 리스트 -> Numpy 배열
X = np.array(X)
y = np.array(y)

print('X.shape:', X.shape) # (17, 3)
print('y.shape:', y.shape)
print('y:', y)
```

```python
X = X.reshape(17, 3, 1)
print(X.shape)
```

```python
# 모델 초기화
model = Sequential([
    Input(shape=(3, 1)),       # 샘플 하나당 3일, 하루당 특성 1개
    SimpleRNN(10),             # RNN 은닉층, 뉴런 10개
    Dense(1)                   # 다음 날 주가 예측, 회귀면 Dense(1), 분류면 Dense(클래스 수)
])
```

```python
# 모델 컴파일
model.compile(optimizer='adam', loss='mse')
```

```python
model.fit(X, y, epochs=100, verbose=0)
```

```python
X_test = price[-3:]
print(X_test)
print(X_test.shape)

X_test = X_test.reshape(-1, 3, 1)
print(X_test)
print(X_test.shape)
```

```python
# 예측 
pred = model.predict(X_test)
print(pred)
```
